```python
import os
import cohere
from dotenv import load_dotenv
```

```python
load_dotenv()
_api_key = os.getenv("COHERE_API_KEY")

if not _api_key:
    raise RuntimeError("Falta COHERE_API_KEY en .env")

co = cohere.ClientV2(_api_key)
GEN_MODEL = "command-a-03-2025"
```

```python
response = co.chat(
        model=GEN_MODEL,
        messages=[{'role':'user', 'content': 'tell me a joke about dolphins'}])
                  
```
